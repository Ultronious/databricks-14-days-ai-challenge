```python
from delta.tables import DeltaTable

deltaTable = DeltaTable.forName(spark, "ecommerce.events")
updatesDF = spark.createDataFrame([], deltaTable.toDF().schema)  # Minimal placeholder

deltaTable.alias("t").merge(
    updatesDF.alias("s"),
    "t.product_id = s.product_id"
).whenMatchedUpdateAll() \
 .whenNotMatchedInsertAll() \
 .execute()
```
